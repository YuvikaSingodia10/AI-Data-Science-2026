```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Aman", "Riya", "Karan", "Neha"],
    "Age": [20, 21, 19, 22],
    "Marks": [85, 92, 78, 88]
})

print(df)
```

```python
df["Name"]
```

```python
#loc is used for label based indexing
df.loc[0]
```

```python
df.loc[1:3]
```

```python
# iloc is used for position based indexing or we can say that index based indexing

df.iloc[0]
```

```python
df.iloc[1:3]
```

```python
# Reset the index first if 'Name' is already the index to make this cell repeatable
if 'Name' not in df.columns:
    df = df.reset_index()

df = df.set_index("Name")
df
```

```python
#conditional selection

df["Marks" ]> 85

df[df["Marks"] > 85]

```

```python
# multiple condition
df[(df["Age"] > 20) & (df["Marks"] > 85)]
```

```python
#####################-----------NUMPY--------------------------

import numpy as np

arr = np.array([10, 20, 30, 40])

print(arr.mean())
print(arr.max())
print(arr.sum())
```

```python
# 2d array in nummpy
arr = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(arr)
```

```python
#shape → rows & columns
#ndim  → number of dimensions
#size  → total elements

arr.shape
print(arr)

arr.ndim
print(arr)

arr.size
print(arr)
```

```python
#slicing in array
arr = np.array([10, 20, 30, 40, 50])

arr = arr[2:4]
print(arr)

print(arr * 2)
```

```python
# statical funtion in numpy
arr = np.array([10, 20, 30, 40, 50])
print(arr.mean())
print(arr.max())
print(arr.sum())
print(np.median(arr))
```
